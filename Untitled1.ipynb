# Executando a Geração de Vídeo (Inference)

Agora você pode carregar seu personagem (uma imagem de referência) e o vídeo contendo os movimentos que deseja copiar. O comando abaixo usa os exemplos padrão do repositório para testar a pipeline de geração.

```python
%cd /content/MimicMotion

import os
import yaml
from huggingface_hub import hf_hub_download

# 1. Garantir que os modelos de detecção de pose (DWPose) estão presentes localmente
os.makedirs("models/DWPose", exist_ok=True)

try:
    print("Baixando arquivos do DWPose (yolox_l.onnx)...")
    hf_hub_download(
        repo_id="yzd-v/DWPose",
        filename="yolox_l.onnx",
        local_dir="models/DWPose"
    )
    print("Baixando arquivos do DWPose (dw-ll_ucoco_384.onnx)...")
    hf_hub_download(
        repo_id="yzd-v/DWPose",
        filename="dw-ll_ucoco_384.onnx",
        local_dir="models/DWPose"
    )
    print("✔ Todos os modelos do DWPose foram baixados!")
except Exception as e:
    print(f"Erro ao baixar modelos DWPose: {e}")

# 2. Corrigir o arquivo mimicmotion/utils/loader.py diretamente usando substituição de texto precisa
loader_path = "mimicmotion/utils/loader.py"
with open(loader_path, "r") as f:
    loader_code = f.read()

target_code = """    with torch.serialization.safe_globals(*allowed_modules):
        mimicmotion_models.load_state_dict(torch.load(ckpt_path, map_location="cpu"))"""

replacement_code = """    try:
        with torch.serialization.safe_globals(*allowed_modules):
            mimicmotion_models.load_state_dict(torch.load(ckpt_path, map_location="cpu"))
    except Exception as e:
        print("[PATCH] Contornando verificação rígida do safe_globals devido ao PyTorch recente...")
        mimicmotion_models.load_state_dict(torch.load(ckpt_path, map_location="cpu", weights_only=False))"""

if target_code in loader_code:
    loader_code = loader_code.replace(target_code, replacement_code)
else:
    loader_code = loader_code.replace("with torch.serialization.safe_globals(*allowed_modules):", "if True:")

with open(loader_path, "w") as f:
    f.write(loader_code)

print("✔ Patch aplicado com sucesso no loader.py!")

# 3. Configurar o arquivo de configuração para usar o modelo público livre de token
with open("configs/test.yaml", "r") as f:
    config = yaml.safe_load(f)

config["base_model_path"] = "vdo/stable-video-diffusion-img2vid-xt-1-1"

with open("configs/test.yaml", "w") as f:
    yaml.dump(config, f)

print("✔ Configuração do modelo base definida para 'vdo/stable-video-diffusion'!")

# 4. Executar o script de inferência
print("Iniciando a inferência... (isso pode levar alguns minutos na primeira execução)")
!python inference.py --inference_config configs/test.yaml
```
